```python
country = "India"  #Global variables
class Student:
   #Class varaibles
    school_name="Kidzee"
    father="Labanu"
    mother="Prasanthi"
    sister="Ishita"

    def __init__(self, name, age):
      #Instance variables
      self.name = name
      self.age = age

    def display_details(self):
      # Local Variables
      message="Student Details"
      print("=======================: "+message)
      print("Name:", self.name)
      print("Age:", self.age)
      print("school:", self.school_name)
      self.display_parentDetails()
        
    def display_parentDetails(self):
      # Local Variables
      message="Family Details"
      print("=======================: "+message)
      print("Father :",self.father)
      print("Mother : ",self.mother)
      print("Sister : ",self.sister)
      print("Country : "+country)
      result=self.dispaly_return()
      print("RESULT : "+result)

    def dispaly_return(self):
      return self.name +" =======>  "+ self.school_name

stu = Student("Raj", 4)
stu.display_details()
```
