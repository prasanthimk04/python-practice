```python
# Operators
# Computers are incredible at doing math calculations.
score = 0           # score is 0
print(score)

score = 4 + 3       # score is now 7
print(score)

score = 4 - 3       # score is now 1
print(score)

score = 4 * 3       # score is now 12
print(score)

score = 4 / 3       # score is now 1.3333
print(score)

# And take a look at this code that calculates a 20% tip by calculating the total and then multiplying by a float (decimal number):
pizza = 2.99
coke = 0.99

tip = (pizza +  coke) * 0.2
print(tip)

# program that converts a number from Fahrenheit (°F) to Celsius (°C).

Fahrenheit = 22
celsius =  (Fahrenheit - 32) / 1.8
print(celsius)

```

```python
#Modulo
score = 5 % 3 # 5 divided by 3 is 1, with a remainder of 2.
print(score)
score = 5 % 2 # 5 divided by 2 is 2, with a remainder of 1.
print(score)
score = 5 % 1 # 5 divided by 1 is 5, with a remainder of 0.
print(score)


# Exponents
# Python can also perform exponentiation such as 2³ or 10².[2 power of 3 or 10 power of 2]
score = 2 ** 2   # score is 4
print(score)
score = 2 ** 3   # score is now 8
print(score)
score = 2 ** 4   # score is now 16
print(score)


#BMI calculates

weight = 64
height = 5.0

bmi = weight / height **  2
print(bmi)
```

```python
# We just got home from a fun trip to South America, specifically Colombia, Peru, and Brazil. There's some leftover cash in:
        # 🇨🇴 Colombian pesos
        # 🇵🇪 Peruvian soles
        # 🇧🇷 Brazilian reais
#Create a program that asks the user for the amount they have in pesos, soles, and reais and calculates the total in USD.

pesos = float(input("how much do you have left in pesos?  "))
soles = float(input("how much do you have left in soles? "))
reais = float(input("how much do you have left in reais? "))

pesos_USD = pesos / 400
soles_USD = soles / 3.5
reais_USD = reais/ 5

total_usd = pesos_USD + soles_USD + reais_USD
print('Total USD remaining', total_usd)
```

```python
# Moore’s Law 
#Moore’s Law states that the number of transistors on a computer chip doubles around every 2 years.
#That means growth isn’t linear – it’s exponential.

# transistors for the number of transistors on your computer right now (e.g. 25000000000).
#years for the time into the future (e.g. 10).


transistors = 2500000000
years = 5

transistors = transistors * 2 ** (years/2)
print(transistors)
```
