```python
import numpy as np
marks = np.array([70,80,99,76, 98])

# Student marks
print("Marks", marks)
print("Average", np.mean(marks))
print("Lowest", np.min(marks))
print("Highest", np.max(marks))
```

```python
import numpy as np

marks = np.array([65, 80, 75, 90, 55, 88, 92])

# 1. Find average
print("Average Marks: ", np.mean(marks))
# 2. Find highest mark
print("highest Marks:", np.max(marks))
# 3. Find lowest mark
print("lowest Marks:", np.min(marks))
# 4. Find marks greater than 80
print("marks greater than 80:", marks > 80)
# 5. Sort the marks
print("Sort:", np.sort(marks))
# 6. Count how many students scored above 80
count = np.sum(marks > 80)
print(count)
```
