```python
# while loop

print('Yes Bank.')

pin = int(input('Enter your PIN: '))

while pin !=1234:
    pin = int(input('Incorrect PIN. Enter your PIN again:'))

if pin == 1234:
    print('Pin accepted!')
```

```python
guess = 0
tries = 0

while guess !=6 and tries <5:
   guess = int(input('Guess the number: ')) 
   tries = tries + 1


if guess !=6:
  print('You ran out of tries.')
else:
  print('You got it!')
```

```python
# FizzBuzz

for num in range(1, 101):
    if num % 3 == 0 and num % 5 == 0:
        print("FizzBuzz")
    elif num % 5 == 0:
        print("Buzz")
    elif num % 3 == 0:
        print("Fizz")
    else:
        print(num)
```

```python
# Nested for loops

for i in range(1, 6):
    for j in range(1, 6):
        print(i * j)
```

```python
i = 0

while i < 6:
  j = 0
  while j < 6:
    print(i * j)
    j = j + 1
    print(j)
  i = i + 1
  print(i)
```

```python
import random

lucky_number = random.randint(1, 9)
not_found = True

while not_found:
  for i in range(1, 10):
    if i == lucky_number:
      not_found = False
      break
    else:
      print(i)

print(f"Yay I got my lucky number {lucky_number}! 🍀")
        
```
