```python
"""
The len() function returns the total length of a list.
The max() function returns the maximum value in a list.
The min() function returns the minimum value in a list.
"""

stock1_prices = [2.52, 2.44, 2.32, 2.41, 2.51, 2.50, 2.44]
stock2_prices = [8.36, 8.31, 8.21, 8.21, 8.25, 8.11, 8.13]
print(len(stock1_prices))
print(max(stock1_prices))
print(min(stock2_prices))

lego_parts = [8980, 7323, 5343, 82700, 92232, 1203, 7319, 8903, 2328, 1279, 679, 589]
print(min(lego_parts))
print(max(lego_parts))
```

# List Method	Description
# .append()	Add an item to the end of the list
# .clear()	Remove all items from the list
# .copy()	Return a shallow copy of the list
# .count()	Return the number of times the value appears in the list
# .extend()	Appends another list to the current list by extending it
# .index()	Returns the index of a value inside the list
# .insert()	Insert an item at a specified position in the list
# .pop()	Remove an item from a specified position in the list
# .remove()	Remove an item from the list based on the value of the item
# .reverse()	Reverses the list in place
# .sort()	Sorts the list in place

```python
# Write code below 💖
"""
.append() method adds an item to the end of the list.
.insert() method adds an item to a specific index.
.remove() method removes an item from a list based on the value.
.pop() method removes the item at a particular index.
"""

dna = ['AUG', 'AUC', 'UCG']
print('Actual list: ', dna)

dna.append('UAA')
print('Append item to the exsting list: ', dna)

dna.insert(2, 'cat')
print('Insert new item in 2nd index:', dna)

dna.remove('cat')
print('Remove cat item')
print(dna)

dna.pop(3)
print('remove index 3 from list',dna)

dna.copy()
print('Return a shallow copy of the list', dna)


print('print count of items that how many times it repeates',dna.count('AUC'))

dna.extend(['cat', 'rat'])
print('Appends another list to the current list by extending it',dna)

print('Returns the index of a value inside the list', dna.index('AUC'))

dna.reverse()
print('Reverses the list in place',dna)

dna.sort()
print('Sorts the list in place',dna)

dna.clear()
print('remve all items',dna)
```

```python
# Iterate Over a List

snowfall = [0.3, 0.0, 0.0, 1.2, 3.9, 2.2, 0.8]

## for-in
for i in snowfall:
  print(i)

# check range and length of list
print('The range and Length of List: ',range(len(snowfall)))


## for-in with range() and len()
for i in range(len(snowfall)):
  print(snowfall[i])


# Theme: Indie Roadtrip 🚗☀️🍃

playlist = [
  'Porches - rangerover',
  'Mount Eerie - You Swan, Go On',
  'Hank Heaven - Threads',
  'Pinegrove - Darkness',
  'LVL UP - Spirit Was',
  'Mitski - First Love / Late Spring'
]

for i in range(len(playlist)):
  print(playlist[i])
```
